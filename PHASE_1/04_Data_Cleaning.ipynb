### Objective
The objective of this notebook is to clean and standardize the Lithium-Ion
Battery Materials dataset.

The cleaning process includes checking duplicate records, standardizing
column names, removing unnecessary spaces from text values, standardizing
categorical values, correcting data types where required, and validating
the dataset after cleaning.

The cleaned dataset will be saved for use in the subsequent data
transformation and exploratory analysis stages.

```python
# ============================================================
# PHASE 1 - DATA CLEANING
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. LOAD DATASET AFTER MISSING VALUE HANDLING
# ------------------------------------------------------------

file_path = "lithium_ion_missing_handled.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("DATA CLEANING")
print("=" * 75)

print("\nORIGINAL DATASET SHAPE:")
print(df.shape)

# ------------------------------------------------------------
# 2. DISPLAY ORIGINAL COLUMN NAMES
# ------------------------------------------------------------

print("\nORIGINAL COLUMN NAMES:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 3. CHECK DUPLICATE RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DUPLICATE RECORD ANALYSIS")
print("=" * 75)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")

print("Shape after duplicate removal:", df.shape)

# ------------------------------------------------------------
# 4. STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("STANDARDIZING COLUMN NAMES")
print("=" * 75)

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace(".", "", regex=False)
    .str.replace("/", "_", regex=False)
)

print("\nCLEANED COLUMN NAMES:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 5. CLEAN TEXT / CATEGORICAL COLUMNS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CLEANING TEXT VARIABLES")
print("=" * 75)

text_columns = df.select_dtypes(include="object").columns.tolist()

for column in text_columns:

    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

print("Text columns cleaned:")
print(text_columns)

# ------------------------------------------------------------
# 6. STANDARDIZE IMPORTANT CATEGORICAL VARIABLES
# ------------------------------------------------------------

# Crystal System
if "crystal_system" in df.columns:
    df["crystal_system"] = (
        df["crystal_system"]
        .str.strip()
        .str.lower()
        .str.title()
    )

# Spacegroup
if "spacegroup" in df.columns:
    df["spacegroup"] = (
        df["spacegroup"]
        .str.strip()
    )

# Formula
if "formula" in df.columns:
    df["formula"] = (
        df["formula"]
        .str.strip()
    )

# Materials ID
if "materials_id" in df.columns:
    df["materials_id"] = (
        df["materials_id"]
        .str.strip()
    )

# ------------------------------------------------------------
# 7. CONVERT NUMERICAL COLUMNS TO NUMERIC DATA TYPES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CHECKING NUMERICAL DATA TYPES")
print("=" * 75)

numerical_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

for column in numerical_columns:

    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

print("\nDATA TYPES AFTER NUMERIC CONVERSION:")
print(df.dtypes)

# ------------------------------------------------------------
# 8. STANDARDIZE BOOLEAN COLUMN
# ------------------------------------------------------------

if "has_bandstructure" in df.columns:

    # Handle possible text representations
    df["has_bandstructure"] = (
        df["has_bandstructure"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "true": True,
            "false": False,
            "1": True,
            "0": False
        })
    )

# ------------------------------------------------------------
# 9. CHECK FOR NEGATIVE / INVALID PHYSICAL VALUES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CHECKING PHYSICAL VALUES")
print("=" * 75)

positive_columns = [
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

for column in positive_columns:

    if column in df.columns:

        negative_count = (df[column] < 0).sum()

        print(
            f"{column}: {negative_count} negative values"
        )

# Formation energy is allowed to be negative,
# therefore it is intentionally not checked as an invalid negative value.

# ------------------------------------------------------------
# 10. CHECK MISSING VALUES AFTER CLEANING
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MISSING VALUE CHECK")
print("=" * 75)

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

display(missing_summary)

# ------------------------------------------------------------
# 11. CHECK DUPLICATES AGAIN
# ------------------------------------------------------------

print("\nDUPLICATES AFTER CLEANING:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 12. DISPLAY CLEANED DATASET
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CLEANED DATASET")
print("=" * 75)

display(df.head())

# ------------------------------------------------------------
# 13. DATASET INFORMATION AFTER CLEANING
# ------------------------------------------------------------

print("\nDATASET INFORMATION AFTER CLEANING:")
df.info()

# ------------------------------------------------------------
# 14. SAVE CLEANED DATASET
# ------------------------------------------------------------

output_file = "lithium_ion_cleaned.csv"

df.to_csv(output_file, index=False)

print("\n" + "=" * 75)
print("CLEANED DATASET SAVED")
print("=" * 75)

print("Output file:", output_file)
print("Final shape:", df.shape)

print("\n" + "=" * 75)
print("DATA CLEANING COMPLETED")
print("=" * 75)
```

# Conclusion

The Lithium-Ion Battery Materials dataset was successfully cleaned and
standardized.

Duplicate records were identified and removed where necessary. Column names
and textual variables were standardized to improve consistency. Numerical
variables were converted to appropriate numeric data types, while the
Boolean bandstructure variable was standardized.

The dataset was also checked for missing values, duplicate records, and
potentially invalid physical values. The cleaned dataset was saved as
`lithium_ion_cleaned.csv`.

The resulting dataset is now ready for data transformation and exploratory
