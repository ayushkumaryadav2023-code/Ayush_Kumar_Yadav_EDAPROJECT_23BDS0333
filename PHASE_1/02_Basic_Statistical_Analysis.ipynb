### Objective
The objective of this notebook is to perform basic statistical analysis
on the numerical and categorical variables of the Lithium-Ion Battery
Materials dataset.

The analysis includes measures of central tendency, measures of dispersion,
minimum and maximum values, range, quartiles, interquartile range,
frequency distributions, and identification of the most frequent categories.

```python
# ============================================================
# PHASE 1 - BASIC STATISTICAL ANALYSIS
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_23BDS0333.csv"
df = pd.read_csv(file_path)

print("=" * 75)
print("BASIC STATISTICAL ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# 2. IDENTIFY NUMERICAL AND CATEGORICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include=["object", "bool"]).columns.tolist()

print("\nNUMERICAL VARIABLES:")
print(numerical_columns)

print("\nCATEGORICAL / BOOLEAN VARIABLES:")
print(categorical_columns)

# ------------------------------------------------------------
# 3. MEAN
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MEAN")
print("=" * 75)

mean_values = df[numerical_columns].mean()
display(mean_values.to_frame("Mean"))

# ------------------------------------------------------------
# 4. MEDIAN
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MEDIAN")
print("=" * 75)

median_values = df[numerical_columns].median()
display(median_values.to_frame("Median"))

# ------------------------------------------------------------
# 5. MODE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MODE")
print("=" * 75)

for column in numerical_columns:
    mode_value = df[column].mode()
    if len(mode_value) > 0:
        print(f"{column}: {mode_value.iloc[0]}")

# ------------------------------------------------------------
# 6. VARIANCE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("VARIANCE")
print("=" * 75)

variance_values = df[numerical_columns].var()
display(variance_values.to_frame("Variance"))

# ------------------------------------------------------------
# 7. STANDARD DEVIATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("STANDARD DEVIATION")
print("=" * 75)

std_values = df[numerical_columns].std()
display(std_values.to_frame("Standard Deviation"))

# ------------------------------------------------------------
# 8. MINIMUM VALUES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MINIMUM VALUES")
print("=" * 75)

min_values = df[numerical_columns].min()
display(min_values.to_frame("Minimum"))

# ------------------------------------------------------------
# 9. MAXIMUM VALUES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MAXIMUM VALUES")
print("=" * 75)

max_values = df[numerical_columns].max()
display(max_values.to_frame("Maximum"))

# ------------------------------------------------------------
# 10. RANGE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("RANGE")
print("=" * 75)

range_values = df[numerical_columns].max() - df[numerical_columns].min()
display(range_values.to_frame("Range"))

# ------------------------------------------------------------
# 11. QUARTILES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("QUARTILES")
print("=" * 75)

quartiles = df[numerical_columns].quantile([0.25, 0.50, 0.75])
display(quartiles)

# ------------------------------------------------------------
# 12. INTERQUARTILE RANGE (IQR)
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("INTERQUARTILE RANGE (IQR)")
print("=" * 75)

Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

display(IQR.to_frame("IQR"))

# ------------------------------------------------------------
# 13. FREQUENCY DISTRIBUTION OF CATEGORICAL VARIABLES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CATEGORICAL FREQUENCY DISTRIBUTIONS")
print("=" * 75)

for column in categorical_columns:
    print(f"\n--- {column} ---")
    display(df[column].value_counts(dropna=False))

# ------------------------------------------------------------
# 14. MOST FREQUENT CATEGORY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MOST FREQUENT CATEGORY")
print("=" * 75)

for column in categorical_columns:
    mode_value = df[column].mode()

    if len(mode_value) > 0:
        print(f"{column}: {mode_value.iloc[0]}")

# ------------------------------------------------------------
# 15. UNIQUE VALUE COUNTS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("NUMBER OF UNIQUE VALUES")
print("=" * 75)

unique_counts = df.nunique()
display(unique_counts.to_frame("Unique Values"))

# ------------------------------------------------------------
# 16. FIVE-NUMBER SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FIVE-NUMBER SUMMARY")
print("=" * 75)

five_number_summary = pd.DataFrame({
    "Minimum": df[numerical_columns].min(),
    "Q1 (25%)": df[numerical_columns].quantile(0.25),
    "Median (50%)": df[numerical_columns].median(),
    "Q3 (75%)": df[numerical_columns].quantile(0.75),
    "Maximum": df[numerical_columns].max()
})

display(five_number_summary)

# ------------------------------------------------------------
# 17. DESCRIPTIVE STATISTICS SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMPLETE DESCRIPTIVE STATISTICS")
print("=" * 75)

display(df[numerical_columns].describe())

# ------------------------------------------------------------
# 18. FINAL SUMMARY TABLE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FINAL STATISTICAL SUMMARY")
print("=" * 75)

summary_table = pd.DataFrame({
    "Mean": df[numerical_columns].mean(),
    "Median": df[numerical_columns].median(),
    "Variance": df[numerical_columns].var(),
    "Std Dev": df[numerical_columns].std(),
    "Minimum": df[numerical_columns].min(),
    "Maximum": df[numerical_columns].max(),
    "Range": df[numerical_columns].max() - df[numerical_columns].min(),
    "IQR": df[numerical_columns].quantile(0.75)
           - df[numerical_columns].quantile(0.25)
})

display(summary_table)

print("\n" + "=" * 75)
print("BASIC STATISTICAL ANALYSIS COMPLETED")
print("=" * 75)
```

Basic statistical analysis was successfully performed on the
Lithium-Ion Battery Materials dataset.

Measures of dispersion including variance, standard deviation, range, and
