```python
# ============================================================
# PHASE 1 - LOADING DATASET
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_23BDS0333.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 70)

# ------------------------------------------------------------
# 2. DISPLAY FIRST 5 RECORDS
# ------------------------------------------------------------

print("\nFIRST 5 RECORDS:")
display(df.head())

# ------------------------------------------------------------
# 3. DISPLAY LAST 5 RECORDS
# ------------------------------------------------------------

print("\nLAST 5 RECORDS:")
display(df.tail())

# ------------------------------------------------------------
# 4. DATASET DIMENSIONS
# ------------------------------------------------------------

print("\nDATASET DIMENSIONS:")
print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

# ------------------------------------------------------------
# 5. COLUMN NAMES
# ------------------------------------------------------------

print("\nCOLUMN NAMES:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

# ------------------------------------------------------------
# 6. DATA TYPES
# ------------------------------------------------------------

print("\nDATA TYPES:")
print(df.dtypes)

# ------------------------------------------------------------
# 7. DATASET INFORMATION
# ------------------------------------------------------------

print("\nDATASET INFORMATION:")
df.info()

# ------------------------------------------------------------
# 8. BASIC STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nNUMERICAL STATISTICAL SUMMARY:")
display(df.describe())

# ------------------------------------------------------------
# 9. COMPLETE STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nCOMPLETE DATASET SUMMARY:")
display(df.describe(include="all"))

# ------------------------------------------------------------
# 10. RANDOM SAMPLE
# ------------------------------------------------------------

print("\nRANDOM SAMPLE OF 5 RECORDS:")
display(df.sample(5, random_state=42))

# ------------------------------------------------------------
# 11. MISSING VALUES
# ------------------------------------------------------------

print("\nMISSING VALUES IN EACH COLUMN:")
missing_values = df.isnull().sum()
display(missing_values.to_frame("Missing Values"))

print("\nMISSING VALUE PERCENTAGE:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
display(missing_percentage.to_frame("Missing Percentage"))

# ------------------------------------------------------------
# 12. DUPLICATE RECORDS
# ------------------------------------------------------------

print("\nNUMBER OF DUPLICATE ROWS:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 13. UNIQUE VALUES IN EACH COLUMN
# ------------------------------------------------------------

print("\nNUMBER OF UNIQUE VALUES IN EACH COLUMN:")

unique_values = df.nunique()
display(unique_values.to_frame("Unique Values"))

# ------------------------------------------------------------
# 14. DATASET MEMORY USAGE
# ------------------------------------------------------------

print("\nMEMORY USAGE:")
print(f"{df.memory_usage(deep=True).sum() / 1024:.2f} KB")

# ------------------------------------------------------------
# 15. FINAL DATASET SHAPE
# ------------------------------------------------------------

print("\nFINAL DATASET SHAPE:")
print(df.shape)

print("\n" + "=" * 70)
print("INITIAL DATASET INSPECTION COMPLETED")
print("=" * 70)
```
