### Objective
The objective of this notebook is to perform univariate exploratory data
analysis on the Lithium-Ion Battery Materials dataset.

Univariate analysis examines one variable at a time to understand its
distribution, central tendency, spread, frequency, and possible outliers.

Histograms, boxplots, and count plots are used to visualize the numerical
and categorical variables in the dataset.

```python
# ============================================================
# PHASE 1 - UNIVARIATE ANALYSIS
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD TRANSFORMED DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("UNIVARIATE ANALYSIS")
print("=" * 75)

print("\nDataset Shape:")
print(df.shape)

# ------------------------------------------------------------
# 2. DISPLAY BASIC INFORMATION
# ------------------------------------------------------------

print("\nDataset Columns:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 3. NUMERICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

# Keep only columns that actually exist
numerical_columns = [
    column for column in numerical_columns
    if column in df.columns
]

print("\nNumerical variables used:")
print(numerical_columns)

# ------------------------------------------------------------
# 4. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DESCRIPTIVE STATISTICS")
print("=" * 75)

display(df[numerical_columns].describe())

# ============================================================
# VISUALIZATION 1
# HISTOGRAM - FORMATION ENERGY
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="formation_energy_ev",
    bins=30,
    kde=True
)

plt.title("Distribution of Formation Energy")
plt.xlabel("Formation Energy (eV)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 2
# HISTOGRAM - BAND GAP
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="band_gap_ev",
    bins=30,
    kde=True
)

plt.title("Distribution of Band Gap")
plt.xlabel("Band Gap (eV)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 3
# BOXPLOT - ENERGY ABOVE HULL
# ============================================================

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    y="e_above_hull_ev"
)

plt.title("Boxplot of Energy Above Hull")
plt.ylabel("Energy Above Hull (eV)")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 4
# BOXPLOT - DENSITY
# ============================================================

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    y="density_gm_cc"
)

plt.title("Boxplot of Material Density")
plt.ylabel("Density (g/cm³)")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 5
# COUNT PLOT - CRYSTAL SYSTEM
# ============================================================

if "crystal_system" in df.columns:

    plt.figure(figsize=(10, 6))

    order = df["crystal_system"].value_counts().index

    sns.countplot(
        data=df,
        y="crystal_system",
        order=order
    )

    plt.title("Distribution of Crystal Systems")
    plt.xlabel("Number of Materials")
    plt.ylabel("Crystal System")
    plt.tight_layout()
    plt.show()

# ============================================================
# VISUALIZATION 6
# COUNT PLOT - BANDSTRUCTURE AVAILABILITY
# ============================================================

if "has_bandstructure" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="has_bandstructure"
    )

    plt.title("Availability of Bandstructure")
    plt.xlabel("Has Bandstructure")
    plt.ylabel("Number of Materials")
    plt.tight_layout()
    plt.show()

# ============================================================
# 5. NUMERICAL VARIABLE SKEWNESS
# ============================================================

print("\n" + "=" * 75)
print("SKEWNESS OF NUMERICAL VARIABLES")
print("=" * 75)

skewness = df[numerical_columns].skew()

display(
    skewness.to_frame("Skewness")
)

# ------------------------------------------------------------
# 6. OUTLIER ANALYSIS USING IQR
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("OUTLIER ANALYSIS USING IQR")
print("=" * 75)

outlier_summary = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Number of Outliers": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)

# ------------------------------------------------------------
# 7. CATEGORICAL FREQUENCY TABLE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CATEGORICAL VARIABLE FREQUENCIES")
print("=" * 75)

categorical_columns = [
    "crystal_system",
    "has_bandstructure"
]

for column in categorical_columns:

    if column in df.columns:

        print(f"\n--- {column} ---")

        frequency_table = df[column].value_counts(
            dropna=False
        )

        display(
            frequency_table.to_frame("Frequency")
        )

print("\n" + "=" * 75)
print("UNIVARIATE ANALYSIS COMPLETED")
print("=" * 75)
```

# Conclusion

Univariate analysis was successfully performed on the Lithium-Ion Battery
Materials dataset.

The distributions of formation energy, band gap, energy above hull, and
density were examined using histograms and boxplots. The distribution of
crystal systems and bandstructure availability was analyzed using count
plots.

Skewness and IQR-based outlier analysis were also performed to understand
the shape and spread of the numerical variables and identify potentially
unusual observations.

The analysis provides an understanding of individual material properties
before investigating relationships between multiple variables.
