### Objective
The objective of this notebook is to perform multivariate exploratory data
analysis on the Lithium-Ion Battery Materials dataset.

Multivariate analysis examines three or more variables simultaneously to
identify relationships, patterns, correlations, and differences among
material properties.

Correlation heatmaps, pair plots, 3D visualizations, grouped analysis,
and multivariable scatter plots are used to understand interactions
between the physical, chemical, and structural properties of the materials.

```python
# ============================================================
# PHASE 1 - MULTIVARIATE ANALYSIS
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from mpl_toolkits.mplot3d import Axes3D
from pandas.plotting import parallel_coordinates

# ------------------------------------------------------------
# 1. LOAD TRANSFORMED DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("MULTIVARIATE ANALYSIS")
print("=" * 75)

print("\nDataset Shape:")
print(df.shape)

# ------------------------------------------------------------
# 2. SELECT IMPORTANT NUMERICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

numerical_columns = [
    column for column in numerical_columns
    if column in df.columns
]

print("\nNumerical variables:")
print(numerical_columns)

# ------------------------------------------------------------
# 3. CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MULTIVARIATE CORRELATION MATRIX")
print("=" * 75)

correlation_matrix = df[numerical_columns].corr()

display(correlation_matrix)

# ============================================================
# VISUALIZATION 1
# CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Material Properties")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 2
# PAIRPLOT
# ============================================================

pairplot_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "density_gm_cc",
    "volume"
]

pairplot_columns = [
    column for column in pairplot_columns
    if column in df.columns
]

pairplot_data = df[pairplot_columns].dropna()

sns.pairplot(
    pairplot_data,
    diag_kind="hist"
)

plt.suptitle(
    "Pairwise Relationships Among Material Properties",
    y=1.02
)

plt.show()

# ============================================================
# VISUALIZATION 3
# 3D SCATTER PLOT
# Formation Energy vs Band Gap vs Density
# ============================================================

fig = plt.figure(figsize=(11, 8))

ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    df["formation_energy_ev"],
    df["band_gap_ev"],
    df["density_gm_cc"],
    c=df["e_above_hull_ev"],
    cmap="viridis",
    s=45,
    alpha=0.75
)

ax.set_title(
    "3D Relationship: Formation Energy, Band Gap and Density"
)

ax.set_xlabel("Formation Energy (eV)")
ax.set_ylabel("Band Gap (eV)")
ax.set_zlabel("Density (g/cm³)")

colorbar = fig.colorbar(
    scatter,
    ax=ax,
    pad=0.1
)

colorbar.set_label("Energy Above Hull (eV)")

plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 4
# MULTIVARIABLE SCATTER PLOT
# ============================================================

plt.figure(figsize=(11, 7))

sns.scatterplot(
    data=df,
    x="formation_energy_ev",
    y="band_gap_ev",
    hue="crystal_system" if "crystal_system" in df.columns else None,
    size="density_gm_cc",
    sizes=(30, 250),
    alpha=0.7
)

plt.title(
    "Formation Energy vs Band Gap with Density and Crystal System"
)

plt.xlabel("Formation Energy (eV)")
plt.ylabel("Band Gap (eV)")

if "crystal_system" in df.columns:
    plt.legend(
        bbox_to_anchor=(1.05, 1),
        loc="upper left"
    )

plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 5
# MULTIVARIATE BOXPLOT
# BAND GAP BY CRYSTAL SYSTEM AND BANDSTRUCTURE
# ============================================================

if (
    "crystal_system" in df.columns
    and "has_bandstructure" in df.columns
):

    plt.figure(figsize=(12, 7))

    sns.boxplot(
        data=df,
        x="crystal_system",
        y="band_gap_ev",
        hue="has_bandstructure"
    )

    plt.title(
        "Band Gap by Crystal System and Bandstructure Availability"
    )

    plt.xlabel("Crystal System")
    plt.ylabel("Band Gap (eV)")
    plt.xticks(rotation=30)

    plt.tight_layout()
    plt.show()

# ============================================================
# VISUALIZATION 6
# PARALLEL COORDINATES
# ============================================================

print("\nPreparing parallel coordinates visualization...")

parallel_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "density_gm_cc",
    "volume"
]

parallel_columns = [
    column for column in parallel_columns
    if column in df.columns
]

parallel_data = df[parallel_columns].copy()

# Normalize variables for comparable scale
for column in parallel_columns:

    minimum = parallel_data[column].min()
    maximum = parallel_data[column].max()

    if maximum != minimum:

        parallel_data[column] = (
            parallel_data[column] - minimum
        ) / (maximum - minimum)

# Create groups using energy above hull
parallel_data["Stability_Group"] = pd.cut(
    df["e_above_hull_ev"],
    bins=[-np.inf, 0.05, 0.15, np.inf],
    labels=["Low", "Medium", "High"]
)

parallel_data = parallel_data.dropna()

plt.figure(figsize=(14, 7))

parallel_coordinates(
    parallel_data,
    "Stability_Group",
    alpha=0.25
)

plt.title(
    "Parallel Coordinates of Normalized Material Properties"
)

plt.ylabel("Normalized Value")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. MULTIVARIATE SUMMARY BY CRYSTAL SYSTEM
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MULTIVARIATE SUMMARY BY CRYSTAL SYSTEM")
print("=" * 75)

if "crystal_system" in df.columns:

    crystal_summary = (
        df.groupby("crystal_system")[numerical_columns]
        .agg(["mean", "median", "std"])
        .round(4)
    )

    display(crystal_summary)

# ------------------------------------------------------------
# 5. MULTIVARIATE SUMMARY BY BANDSTRUCTURE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MULTIVARIATE SUMMARY BY BANDSTRUCTURE")
print("=" * 75)

if "has_bandstructure" in df.columns:

    bandstructure_summary = (
        df.groupby("has_bandstructure")[numerical_columns]
        .mean()
        .round(4)
    )

    display(bandstructure_summary)

# ------------------------------------------------------------
# 6. THREE-VARIABLE STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("THREE-VARIABLE STATISTICAL SUMMARY")
print("=" * 75)

three_variable_columns = [
    "formation_energy_ev",
    "band_gap_ev",
    "density_gm_cc"
]

three_variable_columns = [
    column for column in three_variable_columns
    if column in df.columns
]

display(
    df[three_variable_columns].describe()
)

# ------------------------------------------------------------
# 7. CHECK STRONGEST CORRELATIONS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("STRONGEST VARIABLE RELATIONSHIPS")
print("=" * 75)

correlation_pairs = []

for i in range(len(numerical_columns)):

    for j in range(i + 1, len(numerical_columns)):

        column1 = numerical_columns[i]
        column2 = numerical_columns[j]

        correlation = df[column1].corr(
            df[column2]
        )

        correlation_pairs.append({
            "Variable 1": column1,
            "Variable 2": column2,
            "Correlation": correlation,
            "Absolute Correlation": abs(correlation)
        })

correlation_pairs_df = pd.DataFrame(
    correlation_pairs
)

correlation_pairs_df = correlation_pairs_df.sort_values(
    "Absolute Correlation",
    ascending=False
)

display(correlation_pairs_df)

# ------------------------------------------------------------
# 8. FINAL DATASET CHECK
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FINAL DATASET CHECK")
print("=" * 75)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\n" + "=" * 75)
print("MULTIVARIATE ANALYSIS COMPLETED")
print("=" * 75)
```
