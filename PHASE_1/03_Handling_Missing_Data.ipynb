### Objective
The objective of this notebook is to identify, analyze, and handle missing
values in the Lithium-Ion Battery Materials dataset.

Missing values are first identified using their count and percentage.
Numerical missing values are handled using median imputation, while
categorical and Boolean variables are handled appropriately. The cleaned
dataset is then verified to ensure that missing values have been addressed.

```python
# ============================================================
# PHASE 1 - HANDLING MISSING DATA
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_23BDS0333.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("MISSING DATA ANALYSIS AND HANDLING")
print("=" * 75)

# ------------------------------------------------------------
# 2. DISPLAY ORIGINAL DATASET INFORMATION
# ------------------------------------------------------------

print("\nORIGINAL DATASET SHAPE:")
print(df.shape)

# ------------------------------------------------------------
# 3. IDENTIFY MISSING VALUES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MISSING VALUES BEFORE HANDLING")
print("=" * 75)

missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(missing_summary)

# ------------------------------------------------------------
# 4. TOTAL NUMBER OF MISSING VALUES
# ------------------------------------------------------------

total_missing = df.isnull().sum().sum()

print("\nTOTAL NUMBER OF MISSING VALUES:")
print(total_missing)

# ------------------------------------------------------------
# 5. IDENTIFY NUMERICAL AND CATEGORICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

boolean_columns = df.select_dtypes(include=["bool"]).columns.tolist()

print("\nNUMERICAL COLUMNS:")
print(numerical_columns)

print("\nCATEGORICAL COLUMNS:")
print(categorical_columns)

print("\nBOOLEAN COLUMNS:")
print(boolean_columns)

# ------------------------------------------------------------
# 6. HANDLE MISSING NUMERICAL VALUES
#    MEDIAN IMPUTATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("HANDLING NUMERICAL MISSING VALUES")
print("=" * 75)

for column in numerical_columns:

    missing_before = df[column].isnull().sum()

    if missing_before > 0:

        median_value = df[column].median()

        df[column] = df[column].fillna(median_value)

        print(
            f"{column}: {missing_before} missing values "
            f"replaced with median = {median_value}"
        )

# ------------------------------------------------------------
# 7. HANDLE MISSING CATEGORICAL VALUES
#    USING 'Unknown'
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("HANDLING CATEGORICAL MISSING VALUES")
print("=" * 75)

for column in categorical_columns:

    missing_before = df[column].isnull().sum()

    if missing_before > 0:

        df[column] = df[column].fillna("Unknown")

        print(
            f"{column}: {missing_before} missing values "
            f"replaced with 'Unknown'"
        )

# ------------------------------------------------------------
# 8. HANDLE MISSING BOOLEAN VALUES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("HANDLING BOOLEAN MISSING VALUES")
print("=" * 75)

for column in boolean_columns:

    missing_before = df[column].isnull().sum()

    if missing_before > 0:

        # Convert missing Boolean values to False
        df[column] = df[column].fillna(False)

        print(
            f"{column}: {missing_before} missing values "
            f"replaced with False"
        )

# ------------------------------------------------------------
# 9. VERIFY MISSING VALUES AFTER HANDLING
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MISSING VALUES AFTER HANDLING")
print("=" * 75)

missing_after = df.isnull().sum()

missing_after_percentage = (missing_after / len(df)) * 100

missing_after_summary = pd.DataFrame({
    "Missing Count": missing_after,
    "Missing Percentage": missing_after_percentage
})

display(missing_after_summary)

# ------------------------------------------------------------
# 10. TOTAL MISSING VALUES AFTER HANDLING
# ------------------------------------------------------------

total_missing_after = df.isnull().sum().sum()

print("\nTOTAL MISSING VALUES AFTER HANDLING:")
print(total_missing_after)

# ------------------------------------------------------------
# 11. CHECK DUPLICATE ROWS
# ------------------------------------------------------------

print("\nNUMBER OF DUPLICATE ROWS:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 12. DISPLAY SAMPLE OF PROCESSED DATA
# ------------------------------------------------------------

print("\nSAMPLE OF DATASET AFTER MISSING VALUE HANDLING:")
display(df.head())

# ------------------------------------------------------------
# 13. SAVE PROCESSED DATASET
# ------------------------------------------------------------

output_file = "lithium_ion_missing_handled.csv"

df.to_csv(output_file, index=False)

print("\nPROCESSED DATASET SAVED AS:")
print(output_file)

# ------------------------------------------------------------
# 14. FINAL VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FINAL VERIFICATION")
print("=" * 75)

print("Original Shape :", (len(df), len(df.columns)))
print("Remaining Missing Values :", df.isnull().sum().sum())

if df.isnull().sum().sum() == 0:
    print("STATUS: All missing values have been successfully handled.")
else:
    print("STATUS: Some missing values are still present.")

print("\n" + "=" * 75)
print("MISSING DATA HANDLING COMPLETED")
print("=" * 75)
```

# Conclusion

Missing values in the Lithium-Ion Battery Materials dataset were identified
and analyzed using their counts and percentages.

Numerical missing values were handled using median imputation, while missing
categorical values were replaced with "Unknown". Boolean missing values were
handled using an appropriate Boolean value.

The dataset was then rechecked to verify the remaining missing values and
the processed dataset was saved as `lithium_ion_missing_handled.csv`.

The resulting dataset is now prepared for the data cleaning stage.
