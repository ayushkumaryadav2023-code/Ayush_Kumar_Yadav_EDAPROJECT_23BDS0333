### Objective
The objective of this notebook is to perform bivariate exploratory data
analysis on the Lithium-Ion Battery Materials dataset.

Bivariate analysis examines the relationship between two variables at a time.
The analysis uses correlation, scatter plots, regression plots, boxplots,
and grouped summaries to understand relationships between material
properties and categorical variables.

```python
# ============================================================
# PHASE 1 - BIVARIATE ANALYSIS
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr, linregress

# ------------------------------------------------------------
# 1. LOAD TRANSFORMED DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("BIVARIATE ANALYSIS")
print("=" * 75)

print("\nDataset Shape:")
print(df.shape)

# ------------------------------------------------------------
# 2. SELECT IMPORTANT NUMERICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

numerical_columns = [
    column for column in numerical_columns
    if column in df.columns
]

print("\nNumerical variables:")
print(numerical_columns)

# ------------------------------------------------------------
# 3. CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PEARSON CORRELATION MATRIX")
print("=" * 75)

pearson_corr = df[numerical_columns].corr(method="pearson")

display(pearson_corr)

# ------------------------------------------------------------
# 4. SPEARMAN CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SPEARMAN CORRELATION MATRIX")
print("=" * 75)

spearman_corr = df[numerical_columns].corr(method="spearman")

display(spearman_corr)

# ============================================================
# VISUALIZATION 1
# SCATTER PLOT - FORMATION ENERGY VS ENERGY ABOVE HULL
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="formation_energy_ev",
    y="e_above_hull_ev"
)

plt.title("Formation Energy vs Energy Above Hull")
plt.xlabel("Formation Energy (eV)")
plt.ylabel("Energy Above Hull (eV)")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 2
# REGRESSION PLOT - BAND GAP VS DENSITY
# ============================================================

plt.figure(figsize=(9, 6))

sns.regplot(
    data=df,
    x="band_gap_ev",
    y="density_gm_cc",
    scatter_kws={"alpha": 0.6},
    line_kws={}
)

plt.title("Band Gap vs Density with Regression Line")
plt.xlabel("Band Gap (eV)")
plt.ylabel("Density (g/cm³)")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 3
# SCATTER PLOT - NUMBER OF SITES VS VOLUME
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="nsites",
    y="volume"
)

plt.title("Number of Sites vs Volume")
plt.xlabel("Number of Sites")
plt.ylabel("Volume")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 4
# BOX PLOT - BAND GAP BY CRYSTAL SYSTEM
# ============================================================

if "crystal_system" in df.columns:

    plt.figure(figsize=(11, 6))

    order = (
        df.groupby("crystal_system")["band_gap_ev"]
        .median()
        .sort_values()
        .index
    )

    sns.boxplot(
        data=df,
        x="crystal_system",
        y="band_gap_ev",
        order=order
    )

    plt.title("Band Gap Distribution by Crystal System")
    plt.xlabel("Crystal System")
    plt.ylabel("Band Gap (eV)")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

# ============================================================
# VISUALIZATION 5
# BOX PLOT - DENSITY BY BANDSTRUCTURE AVAILABILITY
# ============================================================

if "has_bandstructure" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.boxplot(
        data=df,
        x="has_bandstructure",
        y="density_gm_cc"
    )

    plt.title("Density by Bandstructure Availability")
    plt.xlabel("Has Bandstructure")
    plt.ylabel("Density (g/cm³)")
    plt.tight_layout()
    plt.show()

# ============================================================
# VISUALIZATION 6
# SCATTER PLOT - FORMATION ENERGY VS BAND GAP
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="formation_energy_ev",
    y="band_gap_ev",
    hue="crystal_system" if "crystal_system" in df.columns else None
)

plt.title("Formation Energy vs Band Gap")
plt.xlabel("Formation Energy (eV)")
plt.ylabel("Band Gap (eV)")

if "crystal_system" in df.columns:
    plt.legend(
        title="Crystal System",
        bbox_to_anchor=(1.05, 1),
        loc="upper left"
    )

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. PAIRWISE CORRELATION DETAILS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PAIRWISE PEARSON CORRELATION")
print("=" * 75)

correlation_results = []

for i in range(len(numerical_columns)):

    for j in range(i + 1, len(numerical_columns)):

        col1 = numerical_columns[i]
        col2 = numerical_columns[j]

        temp = df[[col1, col2]].dropna()

        if len(temp) > 2:

            r, p_value = pearsonr(
                temp[col1],
                temp[col2]
            )

            correlation_results.append({
                "Variable 1": col1,
                "Variable 2": col2,
                "Pearson r": r,
                "p-value": p_value
            })

correlation_df = pd.DataFrame(correlation_results)

display(
    correlation_df.sort_values(
        by="Pearson r",
        key=lambda x: abs(x),
        ascending=False
    )
)

# ------------------------------------------------------------
# 6. SPEARMAN CORRELATION DETAILS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PAIRWISE SPEARMAN CORRELATION")
print("=" * 75)

spearman_results = []

for i in range(len(numerical_columns)):

    for j in range(i + 1, len(numerical_columns)):

        col1 = numerical_columns[i]
        col2 = numerical_columns[j]

        temp = df[[col1, col2]].dropna()

        if len(temp) > 2:

            rho, p_value = spearmanr(
                temp[col1],
                temp[col2]
            )

            spearman_results.append({
                "Variable 1": col1,
                "Variable 2": col2,
                "Spearman rho": rho,
                "p-value": p_value
            })

spearman_df = pd.DataFrame(spearman_results)

display(
    spearman_df.sort_values(
        by="Spearman rho",
        key=lambda x: abs(x),
        ascending=False
    )
)

# ------------------------------------------------------------
# 7. SIMPLE LINEAR REGRESSION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SIMPLE LINEAR REGRESSION")
print("=" * 75)

x_column = "formation_energy_ev"
y_column = "e_above_hull_ev"

regression_data = df[
    [x_column, y_column]
].dropna()

slope, intercept, r_value, p_value, std_err = linregress(
    regression_data[x_column],
    regression_data[y_column]
)

print("Independent Variable :", x_column)
print("Dependent Variable   :", y_column)
print("Slope                :", slope)
print("Intercept            :", intercept)
print("R-squared            :", r_value ** 2)
print("p-value              :", p_value)
print("Standard Error       :", std_err)

# ------------------------------------------------------------
# 8. GROUPED BIVARIATE SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("AVERAGE MATERIAL PROPERTIES BY CRYSTAL SYSTEM")
print("=" * 75)

if "crystal_system" in df.columns:

    grouped_summary = (
        df.groupby("crystal_system")[numerical_columns]
        .mean()
        .round(4)
    )

    display(grouped_summary)

# ------------------------------------------------------------
# 9. FINAL BIVARIATE SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("BIVARIATE ANALYSIS SUMMARY")
print("=" * 75)

print("Pearson correlation analysis completed.")
print("Spearman correlation analysis completed.")
print("Scatter plots and regression analysis completed.")
print("Categorical-numerical relationships analyzed using boxplots.")
print("Grouped numerical summaries generated.")

print("\n" + "=" * 75)
print("BIVARIATE ANALYSIS COMPLETED")
print("=" * 75)
```
