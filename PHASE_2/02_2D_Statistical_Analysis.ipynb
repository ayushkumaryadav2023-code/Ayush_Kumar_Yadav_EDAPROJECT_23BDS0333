### Objective

The objective of this notebook is to perform two-dimensional statistical
analysis on the Lithium-Ion Battery Materials dataset.

The analysis investigates relationships between pairs of numerical variables
using covariance, Pearson correlation, Spearman correlation, scatter plots,
regression analysis, and correlation heatmaps.

Relationships between numerical and categorical variables are also examined
using boxplots and grouped statistical summaries.

```python
# ============================================================
# PHASE 2 - 2D STATISTICAL ANALYSIS
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr, linregress

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("2D STATISTICAL ANALYSIS")
print("=" * 75)

print("\nDataset Shape:")
print(df.shape)

# ------------------------------------------------------------
# 2. SELECT NUMERICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

numerical_columns = [
    column for column in numerical_columns
    if column in df.columns
]

print("\nNumerical variables:")
print(numerical_columns)

# ------------------------------------------------------------
# 3. COVARIANCE MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COVARIANCE MATRIX")
print("=" * 75)

covariance_matrix = df[numerical_columns].cov()

display(covariance_matrix)

# ------------------------------------------------------------
# 4. PEARSON CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PEARSON CORRELATION MATRIX")
print("=" * 75)

pearson_matrix = df[numerical_columns].corr(
    method="pearson"
)

display(pearson_matrix)

# ------------------------------------------------------------
# 5. SPEARMAN CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SPEARMAN CORRELATION MATRIX")
print("=" * 75)

spearman_matrix = df[numerical_columns].corr(
    method="spearman"
)

display(spearman_matrix)

# ============================================================
# VISUALIZATION 1
# SCATTER PLOT
# FORMATION ENERGY VS ENERGY ABOVE HULL
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="formation_energy_ev",
    y="e_above_hull_ev"
)

plt.title("Formation Energy vs Energy Above Hull")
plt.xlabel("Formation Energy (eV)")
plt.ylabel("Energy Above Hull (eV)")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 2
# REGRESSION PLOT
# BAND GAP VS DENSITY
# ============================================================

plt.figure(figsize=(9, 6))

sns.regplot(
    data=df,
    x="band_gap_ev",
    y="density_gm_cc",
    scatter_kws={"alpha": 0.6}
)

plt.title("Band Gap vs Density with Regression Line")
plt.xlabel("Band Gap (eV)")
plt.ylabel("Density (g/cm³)")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 3
# CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(11, 8))

sns.heatmap(
    pearson_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Pearson Correlation Heatmap")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 4
# BOX PLOT
# BAND GAP BY CRYSTAL SYSTEM
# ============================================================

if "crystal_system" in df.columns:

    plt.figure(figsize=(11, 6))

    sns.boxplot(
        data=df,
        x="crystal_system",
        y="band_gap_ev"
    )

    plt.title("Band Gap by Crystal System")
    plt.xlabel("Crystal System")
    plt.ylabel("Band Gap (eV)")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

# ============================================================
# VISUALIZATION 5
# BOX PLOT
# DENSITY BY BANDSTRUCTURE
# ============================================================

if "has_bandstructure" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.boxplot(
        data=df,
        x="has_bandstructure",
        y="density_gm_cc"
    )

    plt.title("Density by Bandstructure Availability")
    plt.xlabel("Has Bandstructure")
    plt.ylabel("Density (g/cm³)")
    plt.tight_layout()
    plt.show()

# ============================================================
# VISUALIZATION 6
# SCATTER PLOT
# NUMBER OF SITES VS VOLUME
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="nsites",
    y="volume"
)

plt.title("Number of Sites vs Volume")
plt.xlabel("Number of Sites")
plt.ylabel("Volume")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 6. PAIRWISE CORRELATION AND SIGNIFICANCE TEST
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PAIRWISE PEARSON CORRELATION TESTS")
print("=" * 75)

pearson_results = []

for i in range(len(numerical_columns)):

    for j in range(i + 1, len(numerical_columns)):

        col1 = numerical_columns[i]
        col2 = numerical_columns[j]

        temp = df[[col1, col2]].dropna()

        if len(temp) > 2:

            r, p_value = pearsonr(
                temp[col1],
                temp[col2]
            )

            pearson_results.append({
                "Variable 1": col1,
                "Variable 2": col2,
                "Pearson r": r,
                "p-value": p_value
            })

pearson_results_df = pd.DataFrame(
    pearson_results
)

display(
    pearson_results_df.sort_values(
        "Pearson r",
        key=lambda x: abs(x),
        ascending=False
    )
)

# ------------------------------------------------------------
# 7. SPEARMAN CORRELATION TESTS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PAIRWISE SPEARMAN CORRELATION TESTS")
print("=" * 75)

spearman_results = []

for i in range(len(numerical_columns)):

    for j in range(i + 1, len(numerical_columns)):

        col1 = numerical_columns[i]
        col2 = numerical_columns[j]

        temp = df[[col1, col2]].dropna()

        if len(temp) > 2:

            rho, p_value = spearmanr(
                temp[col1],
                temp[col2]
            )

            spearman_results.append({
                "Variable 1": col1,
                "Variable 2": col2,
                "Spearman rho": rho,
                "p-value": p_value
            })

spearman_results_df = pd.DataFrame(
    spearman_results
)

display(
    spearman_results_df.sort_values(
        "Spearman rho",
        key=lambda x: abs(x),
        ascending=False
    )
)

# ------------------------------------------------------------
# 8. SIMPLE LINEAR REGRESSION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SIMPLE LINEAR REGRESSION")
print("=" * 75)

x_column = "formation_energy_ev"
y_column = "e_above_hull_ev"

regression_data = df[
    [x_column, y_column]
].dropna()

slope, intercept, r_value, p_value, std_err = linregress(
    regression_data[x_column],
    regression_data[y_column]
)

print("Independent Variable :", x_column)
print("Dependent Variable   :", y_column)
print("Slope                :", slope)
print("Intercept            :", intercept)
print("R-squared            :", r_value ** 2)
print("p-value              :", p_value)
print("Standard Error       :", std_err)

print(
    f"\nRegression Equation:"
    f" {y_column} = "
    f"{intercept:.4f} + "
    f"({slope:.4f} × {x_column})"
)

# ------------------------------------------------------------
# 9. GROUPED TWO-DIMENSIONAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("AVERAGE PROPERTIES BY CRYSTAL SYSTEM")
print("=" * 75)

if "crystal_system" in df.columns:

    grouped_summary = (
        df.groupby("crystal_system")[
            numerical_columns
        ]
        .mean()
        .round(4)
    )

    display(grouped_summary)

# ------------------------------------------------------------
# 10. STRONGEST CORRELATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("STRONGEST PAIRWISE RELATIONSHIP")
print("=" * 75)

if len(pearson_results_df) > 0:

    strongest = pearson_results_df.loc[
        pearson_results_df["Pearson r"].abs().idxmax()
    ]

    print(
        "Variables:",
        strongest["Variable 1"],
        "and",
        strongest["Variable 2"]
    )

    print(
        "Pearson correlation:",
        strongest["Pearson r"]
    )

    print(
        "p-value:",
        strongest["p-value"]
    )

print("\n" + "=" * 75)
print("2D STATISTICAL ANALYSIS COMPLETED")
print("=" * 75)
```

Two-dimensional statistical analysis was successfully performed on the

Covariance, Pearson correlation, and Spearman correlation were calculated
