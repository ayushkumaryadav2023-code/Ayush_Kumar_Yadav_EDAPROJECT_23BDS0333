### Objective

The objective of this notebook is to perform one-dimensional statistical
analysis on the Lithium-Ion Battery Materials dataset.

The analysis includes measures of central tendency, dispersion, percentiles,
skewness, kurtosis, interquartile range, and outlier detection.

Univariate visualizations such as histograms, boxplots, and violin plots
are also used to understand the distribution of individual material
properties.

```python
# ============================================================
# PHASE 2 - 1D STATISTICAL ANALYSIS
# Name        : AYUSH KUMAR YADAV
# Registration: 23BDS0333
# Dataset     : Lithium-Ion Battery Materials
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

file_path = "lithium_ion_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 75)
print("1D STATISTICAL ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# 2. SELECT NUMERICAL VARIABLES
# ------------------------------------------------------------

numerical_columns = [
    "formation_energy_ev",
    "e_above_hull_ev",
    "band_gap_ev",
    "nsites",
    "density_gm_cc",
    "volume"
]

numerical_columns = [
    column for column in numerical_columns
    if column in df.columns
]

print("\nNumerical variables:")
print(numerical_columns)

# ------------------------------------------------------------
# 3. MEAN
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MEAN")
print("=" * 75)

display(
    df[numerical_columns]
    .mean()
    .to_frame("Mean")
)

# ------------------------------------------------------------
# 4. MEDIAN
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MEDIAN")
print("=" * 75)

display(
    df[numerical_columns]
    .median()
    .to_frame("Median")
)

# ------------------------------------------------------------
# 5. MODE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MODE")
print("=" * 75)

for column in numerical_columns:

    mode_value = df[column].mode()

    if len(mode_value) > 0:
        print(
            f"{column}: {mode_value.iloc[0]}"
        )

# ------------------------------------------------------------
# 6. VARIANCE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("VARIANCE")
print("=" * 75)

display(
    df[numerical_columns]
    .var()
    .to_frame("Variance")
)

# ------------------------------------------------------------
# 7. STANDARD DEVIATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("STANDARD DEVIATION")
print("=" * 75)

display(
    df[numerical_columns]
    .std()
    .to_frame("Standard Deviation")
)

# ------------------------------------------------------------
# 8. PERCENTILES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PERCENTILES")
print("=" * 75)

percentiles = df[numerical_columns].quantile(
    [0.10, 0.25, 0.50, 0.75, 0.90]
)

display(percentiles)

# ------------------------------------------------------------
# 9. IQR
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("INTERQUARTILE RANGE")
print("=" * 75)

Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

display(
    IQR.to_frame("IQR")
)

# ------------------------------------------------------------
# 10. SKEWNESS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SKEWNESS")
print("=" * 75)

skewness = df[numerical_columns].skew()

display(
    skewness.to_frame("Skewness")
)

# ------------------------------------------------------------
# 11. KURTOSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("KURTOSIS")
print("=" * 75)

kurtosis = df[numerical_columns].kurtosis()

display(
    kurtosis.to_frame("Kurtosis")
)

# ------------------------------------------------------------
# 12. OUTLIER DETECTION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("OUTLIER DETECTION USING IQR")
print("=" * 75)

outlier_results = []

for column in numerical_columns:

    Q1_value = df[column].quantile(0.25)
    Q3_value = df[column].quantile(0.75)

    IQR_value = Q3_value - Q1_value

    lower_bound = Q1_value - 1.5 * IQR_value
    upper_bound = Q3_value + 1.5 * IQR_value

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({
        "Variable": column,
        "Q1": Q1_value,
        "Q3": Q3_value,
        "IQR": IQR_value,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outliers": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df)

# ============================================================
# VISUALIZATION 1
# HISTOGRAM - FORMATION ENERGY
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="formation_energy_ev",
    bins=30,
    kde=True
)

plt.title("Distribution of Formation Energy")
plt.xlabel("Formation Energy (eV)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 2
# HISTOGRAM - BAND GAP
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="band_gap_ev",
    bins=30,
    kde=True
)

plt.title("Distribution of Band Gap")
plt.xlabel("Band Gap (eV)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 3
# BOXPLOT OF NUMERICAL VARIABLES
# ============================================================

plt.figure(figsize=(12, 7))

scaled_for_plot = df[numerical_columns].copy()

# Standardize only for visualization because
# variables have different units and scales.
for column in numerical_columns:

    mean_value = scaled_for_plot[column].mean()
    std_value = scaled_for_plot[column].std()

    if std_value != 0:
        scaled_for_plot[column] = (
            scaled_for_plot[column] - mean_value
        ) / std_value

sns.boxplot(
    data=scaled_for_plot
)

plt.title("Standardized Boxplots of Numerical Variables")
plt.xlabel("Material Properties")
plt.ylabel("Standardized Value")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# ============================================================
# VISUALIZATION 4
# VIOLIN PLOT - DENSITY
# ============================================================

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    y="density_gm_cc"
)

plt.title("Distribution of Material Density")
plt.ylabel("Density (g/cm³)")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 13. FIVE-NUMBER SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FIVE-NUMBER SUMMARY")
print("=" * 75)

five_number_summary = pd.DataFrame({
    "Minimum": df[numerical_columns].min(),
    "Q1": df[numerical_columns].quantile(0.25),
    "Median": df[numerical_columns].median(),
    "Q3": df[numerical_columns].quantile(0.75),
    "Maximum": df[numerical_columns].max()
})

display(five_number_summary)

# ------------------------------------------------------------
# 14. COMPLETE 1D SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMPLETE 1D STATISTICAL SUMMARY")
print("=" * 75)

summary = pd.DataFrame({
    "Mean": df[numerical_columns].mean(),
    "Median": df[numerical_columns].median(),
    "Variance": df[numerical_columns].var(),
    "Std Dev": df[numerical_columns].std(),
    "Min": df[numerical_columns].min(),
    "Max": df[numerical_columns].max(),
    "IQR": (
        df[numerical_columns].quantile(0.75)
        - df[numerical_columns].quantile(0.25)
    ),
    "Skewness": df[numerical_columns].skew(),
    "Kurtosis": df[numerical_columns].kurtosis()
})

display(summary)

print("\n" + "=" * 75)
print("1D STATISTICAL ANALYSIS COMPLETED")
print("=" * 75)
```

# Conclusion

One-dimensional statistical analysis was successfully performed on the
Lithium-Ion Battery Materials dataset.

Measures of central tendency, dispersion, percentiles, skewness, kurtosis,
and interquartile range were calculated for the numerical material
properties. IQR-based outlier detection was also performed.

Histograms, standardized boxplots, and violin plots were used to visualize
the individual distributions of the material properties.

The analysis provides a detailed understanding of each variable individually
and establishes a foundation for two-dimensional and three-dimensional
